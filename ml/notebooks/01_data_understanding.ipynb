```python
import pandas as pd

df = pd.read_csv("../data/raw/ObesityDataSet_raw_and_data_sinthetic.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
pd.set_option("display.max_columns", None)

df.head()
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].unique())
```

```python
numerical_columns = df.select_dtypes(include="number").columns

print("Numerical columns:")
print(numerical_columns.tolist())

print("\nDescriptive statistics:")
df[numerical_columns].describe().T
```

```python
print("Number of duplicate rows:", df.duplicated().sum())

print("\nPercentage of duplicate rows:",
      round(df.duplicated().mean() * 100, 2), "%")
```

```python
print("Target class distribution:")
print(df["NObeyesdad"].value_counts())

print("\nTarget class distribution (%):")
print(
    (df["NObeyesdad"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

print("Total rows involved in duplicates:", len(duplicate_rows))

print("\nDuplicate rows:")
display(duplicate_rows.sort_values(list(df.columns)))
```

```python
print("Minimum values:")
print(df[numerical_columns].min())

print("\nMaximum values:")
print(df[numerical_columns].max())
```
