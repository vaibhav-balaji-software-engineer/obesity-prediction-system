```python
numerical_columns = df.select_dtypes(include="number").columns
correlation_matrix = df[numerical_columns].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Features")
plt.tight_layout()
plt.show()

print("\nExact correlation values:")
print(correlation_matrix.round(4))
```

```python
# Calculate BMI
df["BMI"] = df["Weight"] / (df["Height"] ** 2)

# Exact BMI statistics by obesity level
bmi_stats = df.groupby("NObeyesdad")["BMI"].agg(
    Count="count",
    Min="min",
    Q1=lambda x: x.quantile(0.25),
    Median="median",
    Q3=lambda x: x.quantile(0.75),
    Max="max",
    Mean="mean"
)

print("Exact BMI values:")
print(bmi_stats)

# Boxplot
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="NObeyesdad",
    y="BMI"
)

plt.title("BMI Distribution Across Obesity Levels")
plt.xlabel("Obesity Level")
plt.ylabel("BMI")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Mean BMI for each obesity class
bmi_means = df.groupby("NObeyesdad")["BMI"].mean().sort_values()

print("Mean BMI by obesity level:")
print(bmi_means)

# Overall BMI range
print("\nOverall BMI range:")
print("Minimum BMI:", df["BMI"].min())
print("Maximum BMI:", df["BMI"].max())

# BMI difference between highest and lowest class means
print("\nDifference between highest and lowest class mean:")
print(bmi_means.max() - bmi_means.min())
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Weight",
    y="BMI",
    hue="NObeyesdad",
    alpha=0.7
)

plt.title("Weight vs BMI")
plt.xlabel("Weight (kg)")
plt.ylabel("BMI")
plt.tight_layout()
plt.show()

# Exact correlation
weight_bmi_corr = df["Weight"].corr(df["BMI"])

print("Exact Weight-BMI correlation:")
print(round(weight_bmi_corr, 4))
```

```python
# BMI distribution across obesity classes
bmi_order = [
    "Insufficient_Weight",
    "Normal_Weight",
    "Overweight_Level_I",
    "Overweight_Level_II",
    "Obesity_Type_I",
    "Obesity_Type_II",
    "Obesity_Type_III"
]

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="NObeyesdad",
    y="BMI",
    order=bmi_order
)

plt.title("BMI Distribution Across Obesity Levels")
plt.xlabel("Obesity Level")
plt.ylabel("BMI")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Exact median BMI for each class
print("Median BMI by obesity level:")
print(
    df.groupby("NObeyesdad")["BMI"]
      .median()
      .reindex(bmi_order)
)
```
